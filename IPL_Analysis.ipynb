```python
import numpy as np

batsman_runs = batsman_runs.astype(int)

unique_batters, inv = np.unique(batter, return_inverse=True)

total_runs = np.zeros(len(unique_batters), dtype=int)
balls_faced = np.zeros(len(unique_batters), dtype=int)

np.add.at(total_runs, inv, batsman_runs)
np.add.at(balls_faced, inv, 1)


strike_rate = np.divide(
    total_runs, balls_faced,
    out=np.zeros_like(total_runs, dtype=float),
    where=balls_faced != 0
) * 100.0


strike_rate_per_batter = [
    (str(unique_batters[i]), float(round(strike_rate[i], 2)))
    for i in range(len(unique_batters))
]

strike_rate_per_batter[:10]
```

```python
import numpy as np

runs_per_ball = batsman_runs.astype(int)

unique_bowlers, inv = np.unique(bowler, return_inverse=True)

runs_conceded = np.zeros(len(unique_bowlers), dtype=int)
balls_bowled = np.zeros(len(unique_bowlers), dtype=int)

np.add.at(runs_conceded, inv, runs_per_ball)
np.add.at(balls_bowled, inv, 1)

overs_bowled = balls_bowled / 6.0

economy_rate = np.divide(
    runs_conceded,
    overs_bowled,
    out=np.zeros_like(runs_conceded, dtype=float),
    where=overs_bowled != 0
)

economy_per_bowler = [
    (str(unique_bowlers[i]), float(round(economy_rate[i], 2)))
    for i in range(len(unique_bowlers))
]

economy_per_bowler
```

```python
import numpy as np

m_ids = match_ids.astype(int)
runs = batsman_runs.astype(int)
teams = batting_team.astype(str)

match_team_pairs = np.column_stack((m_ids, teams))
unique_pairs, inv = np.unique(match_team_pairs, axis=0, return_inverse=True)

team_runs = np.zeros(len(unique_pairs), dtype=int)
np.add.at(team_runs, inv, runs)

unique_matches = np.unique(m_ids)
scorecards = []

for mid in unique_matches:
    mask = unique_pairs[:, 0].astype(int) == mid
    teams_in_match = unique_pairs[mask][:, 1]
    runs_in_match = team_runs[mask]

    if len(teams_in_match) != 2:
        continue

    scorecards.append(
        f"""Match {mid}:
 {teams_in_match[0]}: {int(runs_in_match[0])} runs
 {teams_in_match[1]}: {int(runs_in_match[1])} runs"""
    )

print("\n\n".join(scorecards))
```
