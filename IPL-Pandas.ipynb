```python
deliveries = df.dropna(subset=["match_id", "total_runs"])
```

```python
deliveries["extra_runs"] = deliveries["extra_runs"].fillna(0)
```

```python
deliveries_ids = set(deliveries["match_id"])
```

```python
deliveries_clean = deliveries.copy()

deliveries_clean["extras_type"] = deliveries_clean["extras_type"].fillna("None")
deliveries_clean["dismissal_kind"] = deliveries_clean["dismissal_kind"].fillna("Not Out")
deliveries_clean["player_dismissed"] = deliveries_clean["player_dismissed"].fillna("None")
deliveries_clean["fielder"] = deliveries_clean["fielder"].fillna("None")
```

```python
invalid_wickets = deliveries_clean[
    (deliveries_clean["is_wicket"] == 0) &
    (deliveries_clean["dismissal_kind"] != "Not Out")
]

print("Invalid wicket rows:", len(invalid_wickets))
```

```python
numeric_cols = [
    "inning", "over", "ball", "batsman_runs",
    "extra_runs", "total_runs", "is_wicket"
]

deliveries_clean[numeric_cols] = deliveries_clean[numeric_cols].apply(
    pd.to_numeric, errors="coerce"
)

categorical_cols = [
    "batting_team", "bowling_team",
    "batter", "bowler", "non_striker",
    "extras_type", "dismissal_kind", "fielder"
]

for col in categorical_cols:
    deliveries_clean[col] = deliveries_clean[col].astype("category")
```

```python
deliveries_clean.isna().sum()
```

```python
deliveries_clean.info()
```

```python
def standardize_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )
    return df

deliveries_df = standardize_columns(deliveries_clean)
```

```python
deliveries_df["computed_total_runs"] = (
    deliveries_df["batsman_runs"] + deliveries_df["extra_runs"]
)
```

```python
(deliveries_df["computed_total_runs"] == deliveries_df["total_runs"]).all()
```

```python

deliveries_df["over_ball"] = (
    deliveries_df["over"].astype(str) + "." +
    deliveries_df["ball"].astype(str)
)


deliveries_df["is_wicket"] = deliveries_df["is_wicket"].astype(bool)


deliveries_df["is_legal_delivery"] = deliveries_df["extras_type"].isin(
    ["wide", "no ball"]
) == False
```

```python
matches = pd.read_csv("matches.csv")
```

```python

print("Missing values in matches:")
print(matches.isna().sum())
```

```python
matches["winner"] = matches["winner"].fillna("No Result")
```

```python
matches_ids = set(matches["id"])
```

```python

missing_in_matches = deliveries_ids - matches_ids
print("Match IDs in deliveries but not in matches:", missing_in_matches)
matches_clean = matches.copy()
```

```python
matches_clean = matches.copy()

matches_clean["city"] = matches_clean["city"].fillna("Unknown")
matches_clean["player_of_match"] = matches_clean["player_of_match"].fillna("No Award")
matches_clean["winner"] = matches_clean["winner"].fillna("No Result")
matches_clean["result_margin"] = matches_clean["result_margin"].fillna(0)
matches_clean["target_runs"] = matches_clean["target_runs"].fillna(0)
matches_clean["target_overs"] = matches_clean["target_overs"].fillna(0)
matches_clean["method"] = matches_clean["method"].fillna("Normal")
```

```python
matches_df = standardize_columns(matches_clean)
```

```python

unified_df = deliveries_df.merge(
    matches_df,
    left_on="match_id",
    right_on="id",
    how="inner"
)
```

```python
import pandas as pd

# Helper function for clean display
def section(title):
    print(title.upper())

# 1. Total Runs per Match
section("1. Total Runs per Match")
runs_per_match = unified_df.groupby("match_id")["total_runs"].sum()
display(runs_per_match.head())


# 2. Runs per Team per Match
section("2. Runs per Team per Match")
runs_team_match = (
    unified_df
    .groupby(["match_id", "batting_team"])["total_runs"]
    .sum()
    .reset_index()
)
display(runs_team_match.head())


# 3. Top 10 Batters
section("3. Top 10 Batters by Total Runs")
top_10_batters = (
    unified_df
    .groupby("batter")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
display(top_10_batters)


# 4. Strike Rate of Batters
section("4. Strike Rate of Batters")
strike_rate = (
    unified_df
    .groupby("batter")
    .agg(
        runs=("batsman_runs", "sum"),
        balls=("ball", "count")
    )
)
strike_rate["strike_rate"] = (strike_rate["runs"] / strike_rate["balls"]) * 100
display(strike_rate.sort_values("strike_rate", ascending=False).head(10))


# 5. Top 10 Bowlers by Economy
section("5. Top 10 Bowlers by Economy")
bowling = (
    unified_df
    .groupby("bowler")
    .agg(
        runs_conceded=("total_runs", "sum"),
        balls=("ball", "count")
    )
)
bowling["overs"] = bowling["balls"] / 6
bowling["economy"] = bowling["runs_conceded"] / bowling["overs"]
display(bowling.sort_values("economy").head(10))


# 6. Most Consistent Batters
section("6. Most Consistent Batters")
batter_match_runs = (
    unified_df
    .groupby(["match_id", "batter"])["batsman_runs"]
    .sum()
    .reset_index()
)

consistent = (
    batter_match_runs
    .groupby("batter")
    .agg(
        avg_runs=("batsman_runs", "mean"),
        matches=("match_id", "nunique")
    )
)

display(
    consistent[consistent["matches"] >= 20]
    .sort_values("avg_runs", ascending=False)
    .head(10)
)


# 7. Highest Individual Score in a Match
section("7. Highest Individual Score in a Match")
display(batter_match_runs.sort_values("batsman_runs", ascending=False).head(1))


# 8. Boundary Analysis
section("8. Boundary Analysis")
total_fours = (unified_df["batsman_runs"] == 4).sum()
total_sixes = (unified_df["batsman_runs"] == 6).sum()
print(f"Total Fours: {total_fours}")
print(f"Total Sixes: {total_sixes}")

boundary_players = (
    unified_df
    .assign(
        fours=unified_df["batsman_runs"] == 4,
        sixes=unified_df["batsman_runs"] == 6
    )
    .groupby("batter")[["fours", "sixes"]]
    .sum()
    .sort_values(["sixes", "fours"], ascending=False)
)
display(boundary_players.head(10))


# 9. Boundary Percentage
section("9. Boundary Percentage per Batter")
boundary_pct = (
    unified_df
    .assign(boundary_runs=lambda x: x["batsman_runs"].where(x["batsman_runs"].isin([4, 6]), 0))
    .groupby("batter")
    .agg(
        total_runs=("batsman_runs", "sum"),
        boundary_runs=("boundary_runs", "sum")
    )
)
boundary_pct["boundary_percentage"] = (
    boundary_pct["boundary_runs"] / boundary_pct["total_runs"] * 100
)
display(boundary_pct.sort_values("boundary_percentage", ascending=False).head(10))


# 10. Dot Ball Analysis
section("10. Dot Ball Analysis")
dot_balls = unified_df[unified_df["total_runs"] == 0]
top_dot_bowlers = dot_balls.groupby("bowler").size().sort_values(ascending=False)
display(top_dot_bowlers.head(10))


# 11. Runs per Over Analysis
section("11. Runs per Over Analysis")
runs_per_over = unified_df.groupby("over")["total_runs"].mean()
display(runs_per_over)


# 12. Powerplay Performance (Overs 1–6)
section("12. Powerplay Performance")
powerplay = unified_df[unified_df["over"] <= 6]
display(
    powerplay.groupby("batting_team")["total_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)


# 13. Death Overs Performance (Overs 16–20)
section("13. Death Overs Performance")
death = unified_df[unified_df["over"] >= 16]

display(
    death.groupby("batting_team")["total_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(
    death.groupby("batter")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)


# 14. Run Distribution per Inning
section("14. Run Distribution per Inning")
display(
    unified_df.groupby("inning")["total_runs"]
    .mean()
)


# 15. Toss Impact Analysis
section("15. Toss Impact Analysis")
toss_runs = (
    unified_df
    .groupby(["match_id", "batting_team"])["total_runs"]
    .sum()
    .reset_index()
    .merge(
        unified_df[["match_id", "toss_winner"]].drop_duplicates(),
        on="match_id"
    )
)

toss_runs["is_toss_winner"] = toss_runs["batting_team"] == toss_runs["toss_winner"]
display(toss_runs.groupby("is_toss_winner")["total_runs"].mean())


# 16. Player of Match Contribution
section("16. Player of Match Contribution")
pom = (
    batter_match_runs
    .merge(
        unified_df[["match_id", "player_of_match"]].drop_duplicates(),
        on="match_id"
    )
)

display(pom[pom["batter"] == pom["player_of_match"]].head())


# 17. Venue-wise Analysis
section("17. Venue-wise Analysis")
venue_stats = (
    unified_df
    .groupby("venue")
    .agg(
        matches=("match_id", "nunique"),
        avg_runs=("total_runs", "mean")
    )
)
display(venue_stats.sort_values("avg_runs", ascending=False).head(10))


# 18. City-wise Scoring Trends
section("18. City-wise Scoring Trends")
display(
    unified_df.groupby("city")["total_runs"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)


# 19. Season-wise Run Trends
section("19. Season-wise Run Trends")
display(
    unified_df.groupby("season")["total_runs"]
    .sum()
    .reset_index()
)


# 20. Winning Team Analysis
section("20. Winning Team Analysis")
match_runs = (
    unified_df
    .groupby(["match_id", "batting_team"])["total_runs"]
    .sum()
    .reset_index()
)

predicted = (
    match_runs
    .sort_values(["match_id", "total_runs"], ascending=[True, False])
    .drop_duplicates("match_id")
)

winner_check = predicted.merge(
    unified_df[["match_id", "winner"]].drop_duplicates(),
    on="match_id"
)

winner_check["prediction_correct"] = (
    winner_check["batting_team"] == winner_check["winner"]
)

display(winner_check.head())
```

```python
def snake_case(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )
    return df


def readable_columns(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.replace("_", " ")
        .str.title()
    )
    return df
```

```python
report_runs_per_match = (
    unified_df
    .groupby("match_id")["total_runs"]
    .sum()
    .reset_index()
    .rename(columns={"total_runs": "total_runs_scored"})
    .sort_values("total_runs_scored", ascending=False)
)

report_runs_per_match = readable_columns(report_runs_per_match)
display(report_runs_per_match.head(10))
```

```python
report_team_match_runs = (
    unified_df
    .groupby(["match_id", "batting_team"])["total_runs"]
    .sum()
    .reset_index()
    .rename(columns={
        "batting_team": "team",
        "total_runs": "runs_scored"
    })
    .sort_values(["match_id", "runs_scored"], ascending=[True, False])
)

report_team_match_runs = readable_columns(report_team_match_runs)
display(report_team_match_runs.head(10))
```

```python
report_top_batters = (
    unified_df
    .groupby("batter")["batsman_runs"]
    .sum()
    .reset_index()
    .rename(columns={
        "batter": "player",
        "batsman_runs": "total_runs"
    })
    .sort_values("total_runs", ascending=False)
    .head(10)
)

report_top_batters = readable_columns(report_top_batters)
display(report_top_batters)
```

```python
report_strike_rate = (
    unified_df
    .groupby("batter")
    .agg(
        runs=("batsman_runs", "sum"),
        balls=("ball", "count")
    )
    .reset_index()
)

report_strike_rate["strike_rate"] = (
    report_strike_rate["runs"] / report_strike_rate["balls"] * 100
)

report_strike_rate = (
    report_strike_rate
    .rename(columns={"batter": "player"})
    .sort_values("strike_rate", ascending=False)
)

report_strike_rate = readable_columns(report_strike_rate)
display(report_strike_rate.head(10))
```

```python
report_bowling = (
    unified_df
    .groupby("bowler")
    .agg(
        runs_conceded=("total_runs", "sum"),
        balls=("ball", "count")
    )
    .reset_index()
)

report_bowling["overs"] = report_bowling["balls"] / 6
report_bowling["economy"] = (
    report_bowling["runs_conceded"] / report_bowling["overs"]
)

report_bowling = (
    report_bowling
    .rename(columns={"bowler": "player"})
    .sort_values("economy")
)

report_bowling = readable_columns(report_bowling)
display(report_bowling.head(10))
```

```python
batter_match_runs = (
    unified_df
    .groupby(["match_id", "batter"])["batsman_runs"]
    .sum()
    .reset_index()
)

report_consistency = (
    batter_match_runs
    .groupby("batter")
    .agg(
        avg_runs_per_match=("batsman_runs", "mean"),
        matches_played=("match_id", "nunique")
    )
    .reset_index()
)

report_consistency = (
    report_consistency
    .rename(columns={"batter": "player"})
    .sort_values("avg_runs_per_match", ascending=False)
)

report_consistency = readable_columns(report_consistency)
display(report_consistency.head(10))
```

```python
report_venues = (
    unified_df
    .groupby("venue")
    .agg(
        matches=("match_id", "nunique"),
        avg_runs_per_ball=("total_runs", "mean")
    )
    .reset_index()
    .sort_values("avg_runs_per_ball", ascending=False)
)

report_venues = readable_columns(report_venues)
display(report_venues.head(10))
```
